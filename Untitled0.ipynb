```python
import random
import numpy as np

# Sphere Objective Function
def sphere(individual):
    return sum(x**2 for x in individual)

# Optimization parameters
POP_SIZE = 50
GENERATIONS = 50
LOWER_BOUND, UPPER_BOUND = -5.12, 5.12

def genetic_algorithm_sphere():
    # Initialize population
    pop = [[random.uniform(LOWER_BOUND, UPPER_BOUND) for _ in range(2)] for _ in range(POP_SIZE)]

    for gen in range(GENERATIONS):
        pop.sort(key=sphere)
        best_fit = sphere(pop[0])

        # Elitism (Keep top 2)
        new_pop = [pop[0].copy(), pop[1].copy()]

        while len(new_pop) < POP_SIZE:
            # Tournament selection (Minimization)
            p1 = min(random.sample(pop, 3), key=sphere)
            p2 = min(random.sample(pop, 3), key=sphere)

            # Arithmetic Crossover
            alpha = random.random()
            c1 = [alpha * p1[i] + (1 - alpha) * p2[i] for i in range(2)]

            # Gaussian Mutation
            for i in range(2):
                if random.random() < 0.2:
                    c1[i] += random.gauss(0, 0.1)
                    c1[i] = np.clip(c1[i], LOWER_BOUND, UPPER_BOUND)

            new_pop.append(c1)

        pop = new_pop

    pop.sort(key=sphere)
    return pop[0], sphere(pop[0])

best_sol, best_val = genetic_algorithm_sphere()
print(f"Optimal Point Found: x1 = {best_sol[0]:.6f}, x2 = {best_sol[1]:.6f}")
print(f"Minimum Sphere Value: {best_val:.8f}")
```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

# Objective function: 2D Sphere Benchmark (Minimization)
def sphere(individual):
    x1 = individual[0]
    x2 = individual[1]
    return x1**2 + x2**2

# GA Parameters
POP_SIZE = 30
GENERATIONS = 30
LOWER_BOUND = -5.12
UPPER_BOUND = 5.12
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.2
MUTATION_STD = 0.5
TOURNAMENT_SIZE = 3
DIMENSIONS = 2
ELITE_SIZE = 1

# Create random initial population
def create_population():
    population = []
    for _ in range(POP_SIZE):
        individual = [
            random.uniform(LOWER_BOUND, UPPER_BOUND)
            for _ in range(DIMENSIONS)
        ]
        population.append(individual)
    return population

# Tournament Selection (Minimization)
def tournament_selection(population):
    competitors = random.sample(population, TOURNAMENT_SIZE)
    competitors.sort(key=sphere)
    return competitors[0].copy()

# Arithmetic Crossover
def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        alpha = random.random()
        child1 = [alpha * parent1[i] + (1 - alpha) * parent2[i] for i in range(DIMENSIONS)]
        child2 = [alpha * parent2[i] + (1 - alpha) * parent1[i] for i in range(DIMENSIONS)]
        return child1, child2
    return parent1.copy(), parent2.copy()

# Gaussian Mutation with Boundary Clamping
def mutation(individual):
    for i in range(DIMENSIONS):
        if random.random() < MUTATION_RATE:
            individual[i] += random.gauss(0, MUTATION_STD)
            individual[i] = np.clip(individual[i], LOWER_BOUND, UPPER_BOUND)
    return individual

# Genetic Algorithm Execution Pipeline
def genetic_algorithm():
    population = create_population()
    best_fitness_history = []
    avg_fitness_history = []

    for gen in range(GENERATIONS):
        # Rank population by fitness
        population.sort(key=sphere)

        best_ind = population[0].copy()
        best_fit = sphere(best_ind)
        avg_fit = np.mean([sphere(ind) for ind in population])

        best_fitness_history.append(best_fit)
        avg_fitness_history.append(avg_fit)

        # Elitism: Retain top individual
        new_population = [best_ind]

        # Breeding loop
        while len(new_population) < POP_SIZE:
            p1 = tournament_selection(population)
            p2 = tournament_selection(population)
            c1, c2 = crossover(p1, p2)
            c1 = mutation(c1)
            c2 = mutation(c2)

            new_population.append(c1)
            if len(new_population) < POP_SIZE:
                new_population.append(c2)

        population = new_population

    population.sort(key=sphere)
    return population[0], sphere(population[0]), best_fitness_history, avg_fitness_history

# Execution & Output
random.seed(42)
np.random.seed(42)

best_solution, best_fitness, best_hist, avg_hist = genetic_algorithm()

print("=" * 45)
print("SPHERE FUNCTION GA OPTIMIZATION RESULTS")
print("=" * 45)
print(f"Optimal Coordinates: x1 = {best_solution[0]:.6f}, x2 = {best_solution[1]:.6f}")
print(f"Minimum Fitness f(x1, x2): {best_fitness:.8f}")
print("=" * 45)

# Plotting Convergence
plt.figure(figsize=(8, 5))
plt.plot(best_hist, label="Best Fitness", color="blue", linewidth=2)
plt.plot(avg_hist, label="Average Population Fitness", color="orange", linestyle="--")
plt.xlabel("Generation")
plt.ylabel("Sphere Value f(x1, x2)")
plt.title("GA Convergence Curve on Sphere Function")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)


# 1. Define objective function (Sphere function: f(x) = sum(x^2))
def objective_function(x):
    return np.sum(x**2)


# 2. Genetic Algorithm function returning best fitness per generation
def run_genetic_algorithm(
    generations=50, pop_size=30, num_vars=2, bounds=(-5, 5)
):
    # Initialize random population
    population = np.random.uniform(
        bounds[0], bounds[1], size=(pop_size, num_vars)
    )

    best_fitness_history = []

    for gen in range(generations):
        # Evaluate fitness (lower is better for minimization)
        fitness = np.array([objective_function(ind) for ind in population])

        # Track best fitness in current generation
        best_fitness_history.append(np.min(fitness))

        # Tournament selection
        selected = []
        for _ in range(pop_size):
            i, j = np.random.choice(pop_size, 2, replace=False)
            selected.append(
                population[i] if fitness[i] < fitness[j] else population[j]
            )
        population = np.array(selected)

        # Crossover & Mutation (Arithmetic crossover + Gaussian mutation)
        for i in range(0, pop_size - 1, 2):
            if np.random.rand() < 0.8:
                alpha = np.random.rand()
                parent1, parent2 = population[i].copy(), population[i + 1].copy()
                population[i] = alpha * parent1 + (1 - alpha) * parent2
                population[i + 1] = (1 - alpha) * parent1 + alpha * parent2

        # Mutation
        mutation_mask = np.random.rand(*population.shape) < 0.1
        population += mutation_mask * np.random.normal(
            0, 0.2, size=population.shape
        )
        population = np.clip(
            population, bounds[0], bounds[1]
        )  # Enforce bounds

    return best_fitness_history


# 3. Perform 30 Independent Runs
NUM_RUNS = 30
GENERATIONS = 50

all_runs_history = []
final_fitness_values = []

for run in range(NUM_RUNS):
    history = run_genetic_algorithm(generations=GENERATIONS)
    all_runs_history.append(history)
    final_fitness_values.append(history[-1])

all_runs_history = np.array(all_runs_history)
final_fitness_values = np.array(final_fitness_values)

# 4. Calculate Statistical Metrics
stats = {
    "Mean": np.mean(final_fitness_values),
    "Median": np.median(final_fitness_values),
    "Std Dev": np.std(final_fitness_values),
    "Min (Best)": np.min(final_fitness_values),
    "Max (Worst)": np.max(final_fitness_values),
}

# Print Summary Statistics
print("=== Statistical Analysis over 30 Independent Runs ===")
for key, value in stats.items():
    print(f"{key:12s}: {value:.6e}")

# 5. Visualizations
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Convergence across all runs + Mean curve
for run in range(NUM_RUNS):
    axes[0].plot(all_runs_history[run], color="gray", alpha=0.3, linewidth=1)
axes[0].plot(
    np.mean(all_runs_history, axis=0),
    color="red",
    linewidth=2,
    label="Mean Convergence",
)
axes[0].set_xlabel("Generation")
axes[0].set_ylabel("Best Fitness")
axes[0].set_title("GA Convergence Across 30 Runs")
axes[0].set_yscale("log")
axes[0].grid(True, which="both", linestyle="--", alpha=0.5)
axes[0].legend()

# Plot 2: Boxplot of final fitness values
axes[1].boxplot(final_fitness_values, patch_artist=True, boxprops=dict(facecolor="lightblue"))
axes[1].set_ylabel("Final Fitness Score")
axes[1].set_title("Distribution of Final Fitness (30 Runs)")
axes[1].set_yscale("log")
axes[1].grid(True, which="both", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
